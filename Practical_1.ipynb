```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler, MinMaxScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.feature_selection import SelectKBest, f_regression
from sklearn.decomposition import PCA
```

```python
df = pd.read_csv("/content/auto-mpg.csv")
display(df.head())
```

```python
print(df.dtypes)
```

```python
numeric_column = df.select_dtypes(include = np.number)
```

```python
df.select_dtypes(include = np.number).columns
```

```python
df.select_dtypes(include= "object").columns
```

```python
df["origin"] = df["origin"].astype("object")
```

```python
df = df.replace("?" , np.nan)
print(df.isnull().sum())
```

```python
df["horsepower"] = pd.to_numeric(df["horsepower"])

df["horsepower"] = df["horsepower"].fillna(df["horsepower"].mean())
print(df.isnull().sum())
```

```python
scaler = MinMaxScaler().set_output(transform = "pandas")
scaled_data = scaler.fit_transform(numeric_column)
scaled_data.head()
```

```python
encoder = OneHotEncoder(
    handle_unknown= "ignore",
    sparse_output= False).set_output(transform = "pandas")

encoded_data = encoder.fit_transform(df[["origin"]])
display(encoded_data.head())
```

```python
X = df.drop("mpg", axis=1)
y = df["mpg"]
```

```python
X_numeric = X.select_dtypes(include = np.number)

selector = SelectKBest(score_func= f_regression, k = 3).set_output(transform = "pandas")
X_selected = selector.fit_transform(X_numeric, y)
print(X_selected.head())
```
